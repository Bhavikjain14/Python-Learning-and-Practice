2. Circumference of Circle
3. Volume of Cube
4. Volume of Cuboid
5. Exit
Take two numbers as input from the user.
Display the above menu.
Perform the selected operation using if-elif-else.
For options where only one value is needed, use the first number and ignore the second number.

```python
print("1 for Perimeter of Rectangle")
print("2 for Circumference of Circle")
print("3 for Volume of Cube")
print("4 for Volume of Cuboid")
print("5 for Exit")

l = int(input("Enter length "))
b = int(input("Enter breadth "))
r = int(input("Enter radius "))
s = int(input("Enter side "))

ch=int(input("Enter your choice "))

if(ch==1):
  Result = 2*(l+b)

elif(ch==2):
  Result = 2*3.14*r

elif(ch==3):
  Result = s*s*s

elif(ch==4):
  Result = l*b*s

else:
  Result=0
print("Result",Result)
```

```python
# w a p take two numbers from the user then if number is % by 3 then double this number otherwise half this number

num1 = int(input("Enter a no "))
num2 = int(input("Enter b no "))

for i in range(num1, num2+1, 1):
  if(i%3==0):
    print(i*2)
  else:
    print(i/2)
```
