```python
from typing import ValuesView
import numpy as np
s=[100,np.nan,300,400,500,np.nan,700]
day=["Mon","Tues","Wed","Thur","Fri","Sat","Sun"]
Sales=pd.Series(s,index=day)

print(Sales["Tues"])

print(Sales)

print(Sales.isnull())               # Give trues and false Values

print(Sales.isnull().sum())         # give count of NULL Values

print(Sales.notnull())             # Give trues and false Values

print(Sales.notnull().sum())        # give count of NOT NULL Values

print(Sales.dropna())              # Drop NULL Values

print(Sales.fillna(0))            # Fill the null values with bracket value

print(Sales.fillna(Sales.min()))   # fill the null values by min of sales figures

print(Sales.fillna(Sales.max()))   # fill the null values by max of sales figures

print(Sales.fillna(Sales.mean()))  # fill the null values by average of sales figures
```

```python
import pandas as pd
customers = {
    "Customer_ID": [101, 102, 103, 104, 105],
    "Customer_Name": ["Rahul", "Priya", "Amit", "Neha", "Rohit"],
    "City": ["Delhi", "Mumbai", "Pune", "Delhi", "Jaipur"]
}
customer_df = pd.DataFrame(customers)

orders = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006],
    "Customer_ID": [101, 102, 101, 103, 105, 102],
    "Product": ["Laptop", "Mobile", "Mouse", "Keyboard", "Monitor", "Headphones"],
    "Amount": [55000, 30000, 1200, 2500, 15000, 3000]
}
order_df = pd.DataFrame(orders)

print(customer_df)
print(order_df)

Rdf=pd.merge(customer_df,order_df,on="Customer_ID")
print(Rdf)

sdf=pd.merge(customer_df,order_df,on="Customer_ID", how="left")
print(sdf)

tdf=pd.merge(customer_df,order_df,on="Customer_ID", how="right")
print(tdf)

udf=pd.merge(customer_df,order_df,on="Customer_ID", how="outer")
print(udf)
```

```python
import pandas as pd

data = {
    "Order_ID": [101,102,103,104,105,106,107,108,109,110,
                 111,112,113,114,115,116,117,118,119,120],

    "Product": ["Laptop","Mobile","Headphones","Keyboard","Mouse",
                "Laptop","Mobile","Monitor","Tablet","Headphones",
                "Keyboard","Mouse","Laptop","Mobile","Monitor",
                "Tablet","Headphones","Keyboard","Mouse","Laptop"],

    "Category": ["Electronics","Electronics","Accessories","Accessories","Accessories",
                 "Electronics","Electronics","Electronics","Electronics","Accessories",
                 "Accessories","Accessories","Electronics","Electronics","Electronics",
                 "Electronics","Accessories","Accessories","Accessories","Electronics"],

    "City": ["Delhi","Mumbai","Delhi","Pune","Mumbai",
             "Delhi","Bangalore","Pune","Hyderabad","Mumbai",
             "Delhi","Bangalore","Hyderabad","Pune","Delhi",
             "Mumbai","Bangalore","Hyderabad","Pune","Delhi"],

    "Quantity": [2,1,3,2,4,1,2,1,2,3,
                 2,5,1,2,3,1,4,2,3,1],

    "Price": [55000,30000,2000,1500,800,
              60000,28000,22000,35000,2500,
              1800,700,58000,32000,21000,
              36000,2200,1600,750,62000]
}

df = pd.DataFrame(data)

print(df.groupby("City")["Price"].max())

print(df.groupby("Category")["Quantity"].count())

print(pd.pivot_table(df,index="City",columns="Category",values="Price",aggfunc="sum"))        # by default it will take mean values if aggfunc not used

print(pd.pivot_table(df,index="Product", columns="City",values="Quantity", aggfunc="max"))
```
