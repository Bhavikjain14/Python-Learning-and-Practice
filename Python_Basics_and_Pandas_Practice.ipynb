```python
import pandas as pd
products = {
    "orderId": [
        101, 102, 103, 104, 105,
        106, 107, 108, 109, 110,
        111, 112, 113, 114, 115,
        116, 117, 118, 119, 120
    ],
    "Name": [
        "pen", "notebook", "Laptop", "Mobile", "watch",
        "keyboard", "mouse", "tablet", "bag", "headphone",
        "printer", "monitor", "charger", "speaker", "chair",
        "desk", "bottle", "calculator", "diary", "camera"
    ],
    "qty": [
        200, 100, 30, 50, 90,
        80, 120, 40, 70, 60,
        25, 35, 100, 45, 20,
        15, 110, 55, 75, 18
    ],
    "city": [
        "pune", "delhi", "pune", "mumbai", "mumbai",
        "delhi", "pune", "delhi", "mumbai", "pune",
        "delhi", "mumbai", "pune", "delhi", "mumbai",
        "pune", "delhi", "mumbai", "pune", "delhi"
    ],
    "cost": [
        100, 3000, 5000, 600, 9000,
        1500, 800, 25000, 1200, 3000,
        15000, 12000, 700, 2500, 8000,
        10000, 500, 400, 300, 45000
    ]
}
df = pd.DataFrame(products)

print(df.head())        #by default first five rows
                                                                #we can also insert values in ()
print(df.tail())        #by default last five rows

print(df.shape)         #matrix (rows, comuns)

print(df.columns)       #provide column names

print(df.info())        #provide information of the dataset

print(df.describe())    #work on numeric values and orovide statistical data

print(df[['Name',"city"]])  #provide only name and city column

print(df[["Name", "city"]]. head())   #provide only name and city column and five rows

print(df.loc[6])      #only provide data for 6th row

print(df.loc[6:11])    #provide data in rnage of 6 and 11

print(df.loc[[3,6,9]])    #provide data present in 3,6,9 rows

print(df[df['Name'] == 'Laptop'])   #provide only for laptops

print(df[df["qty"]>100])      #provide only for qty greater than 100

print(df[df["cost"]<9000])    #provide only for cost of product less than 9000

print(df[(df['city']=="pune") | (df['city']=="mumbai")])    # | work as "or"

print(df[(df['Name']=="Mobile") & (df['qty']>=50)])    # & work as "and"

print(df["Name"])
# Filter the DataFrame for costs between 500 and 10000 (inclusive) and print the 'Name' column
filtered_names = df[(df["cost"] >= 500) & (df["cost"] <= 10000)]
print(filtered_names["Name"])
```
